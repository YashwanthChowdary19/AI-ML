## <b>What is File Handling?
    Definition: File handling is the process of creating, opening, reading, writing, and managing files using Python.
    
    Key Point: Backend systems constantly interact with files—logs, configs, uploads, and temporary storage.
    
    Analogy: Think of a file as a notebook. Opening it is like flipping to a page, reading is like scanning notes, writing is like adding new text, and closing is like putting the notebook back on the shelf.


### example :
    logs.txt
    config.json
    users.csv
    uploads/image.png

### <b>4. File Modes

| Mode | Meaning           |
| ---- | ----------------- |
| r    | read              |
| w    | write (overwrite) |
| a    | append            |
| x    | create new file   |
| rb   | read binary       |
| wb   | write binary      |
| r+   | read and write    |
| w+   | write and read    |
| a+   | append and read   |
