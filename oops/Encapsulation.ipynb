### <b>What is Encapsulation?
    Encapsulation means wrapping data and methods together inside a class and restricting direct access to some/senstive data.

### <b>Simple Real-World Example
    Think of a Capsule (medicine)

        Inside → medicine (data)

        Outside → capsule cover (protection)

    You cannot directly touch the medicine.
    You must consume it properly.

    That protection = Encapsulation

### <b>Banking Example

```python
class BankAccount: 
    def __init__(self,balance=0): 
        self.balance = balance 
```

    If balance is public:

```python
account1 = BankAccount()
```

```python
account1.balance=1000  # ❌ Dangerous
```

```python
account1.balance
```

    Anyone can change balance.
    That’s BAD design.

### <b> How to Fix ? 
    solution is make the balance private 

```python
class BankAccount: 
    def __init__(self:object,balance:float):
        self.__balance = balance

account1 = BankAccount(1000)    
account1.__balance
```

    now Access is restricted. so they can't modify

### <b>Types of Access in Python

    Python doesn’t have strict access modifiers like Java,
    but it follows naming conventions.

#### <b>1️⃣ Public (Default)
    self.balance

Accessible everywhere.

#### <b>2️⃣ Protected (Single Underscore)
    self._balance
Accessible within the class and sub class

#### <b>3️⃣ Private (Double Underscore)
    self.__balance 
Accessible within the class itself

### <b>Important Concept: Name Mangling 
    Name mangling is the process where Python internally changes a private attribute name from __variable to _ClassName__variable to avoid accidental access or overriding in subclasses.
    
    Name mangling does not make variables truly private. It only prevents accidental access, not intentional access.
    
    Python is saying:

        “I won’t completely block access, but I will make it harder so you don’t access it accidentally.”

```python
# example 
class BankAccount: 
    def __init__(self,balance): 
        self.__balance = balance

account = BankAccount(10000)
account._BankAccount__balance # it is working even though it is private attribute due to name mangling
```

### <b>Why Encapsulation is Important

    ✔ Protects sensitive data
    ✔ Prevents accidental modification
    ✔ Controls how data changes
    ✔ Makes code maintainable
