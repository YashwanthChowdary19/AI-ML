### <b>What is Encapsulation?
    Encapsulation means wrapping data and methods together inside a class and restricting direct access to some/senstive data.

### <b>Simple Real-World Example
    Think of a Capsule (medicine)

        Inside → medicine (data)

        Outside → capsule cover (protection)

    You cannot directly touch the medicine.
    You must consume it properly.

    That protection = Encapsulation

### <b>Banking Example

```python
class BankAccount: 
    def __init__(self,balance=0): 
        self.balance = balance 
```

    If balance is public:

```python
account1 = BankAccount()
```

```python
account1.balance=1000  # ❌ Dangerous
```

```python
account1.balance
```

    Anyone can change balance.
    That’s BAD design.

### <b> How to Fix ? 
    solution is make the balance private 

```python
class BankAccount: 
    def __init__(self:object,balance:float):
        self.__balance = balance

account1 = BankAccount(1000)    
account1.__balance
```

    now Access is restricted. so they can't modify

### <b>Types of Access in Python

    Python doesn’t have strict access modifiers like Java,
    but it follows naming conventions.

#### <b>1️⃣ Public (Default)
    self.balance

Accessible everywhere.

#### <b>2️⃣ Protected (Single Underscore)
    self._balance
Accessible within the class and sub class

#### <b>3️⃣ Private (Double Underscore)
    self.__balance 
Accessible within the class itself

### <b>Important Concept: Name Mangling 
    Name mangling is the process where Python internally changes a private attribute name from __variable to _ClassName__variable to avoid accidental access or overriding in subclasses.
    
    Name mangling does not make variables truly private. It only prevents accidental access, not intentional access.
    
    Python is saying:

        “I won’t completely block access, but I will make it harder so you don’t access it accidentally.”

```python
# example 
class BankAccount: 
    def __init__(self,balance): 
        self.__balance = balance

account = BankAccount(10000)
account._BankAccount__balance # it is working even though it is private attribute due to name mangling
```

### <b>Why Encapsulation is Important

    ✔ Protects sensitive data
    ✔ Prevents accidental modification
    ✔ Controls how data changes
    ✔ Makes code maintainable

## <b>Getter and Setter Method

### <b>1️⃣ Why Do We Need Getter & Setter?

    imagine:

```python
class User: 
    def __init__(self, age:int): 
        self.age = age 
```

    Now anyone can do:

```python
us = User(20)
```

```python
us.age = 80 
us.age
```

    That’s bad.
    
    We need control over how values are set.
    
    That control = Getter & Setter.

### <b>2️⃣ Traditional Getter & Setter (Old Style)

```python
class User: 
    def __init__(self,age): 
        self.__age = age 

    def get_age(self): 
        return self.__age

    def set_age(self,new_age):
        if new_age > 0: 
            self.__age = new_age 
        else : 
            return "Invalid age"
        
```

```python
u1 = User(20)
u1.get_age()
```

```python
u1.set_age(-10)
```

```python
u1.set_age(80)
```

```python
u1.get_age()
```

### <b>❌ Problem With Traditional Approach

    usage looks ugly: 
    u.get_age()
    u.set_age(20)
    
    Instead of : 
    u.age
    u.age = 10

### <b>Python solves this using @property.

- This is the modern and correct way.

#### <b>Step 1: Create Getter

```python
class User: 
    def __init__(self,age): 
        self.__age = age

    @property 
    def age(self) : 
        return self.__age

u1 = User(20)
u1.age
```

- Looks like attribute access
- But actually calls method.

```python
# Explanation : 
- for getters we can @property decorator 
```

### <b>Step 2: Create Setter


```python
class User: 
    def __init__(self, age): 
        self.__age = age 

    @property 
    def age(self): 
        return self.__age
        
    @age.setter 
    def age(self, new_age): 
        if new_age > 0: 
            self.__age = new_age 
        else: 
            raise ValueError("Invalid age") 
u1 = User(20)
```

```python
u1.age = 30   # ✅
u1.age = -5   # ❌ Invalid age
```

    ✔ Clean
    ✔ Controlled
    ✔ Professional

### <b>What Is Happening Internally?

```python
# When you do: 
u1.age
# Python calls:
User.age(u1)
```

```python
# When you do: 
u1.age = 20 
# Python calls:
User.age.setter(u1, 30)
```

    So it looks like variable access, but it’s actually method control.

### <b>4️⃣ Deleter (Optional)

```python
class User: 
    def __init__(self, age): 
        self.__age = age 

    @property 
    def age(self): 
        return self.__age
        
    @age.setter 
    def age(self, new_age): 
        if new_age > 0: 
            self.__age = new_age 
        else: 
            raise ValueError("Invalid age")

    @age.deleter 
    def age(self): 
        print("deleting age")
        del self.__age
        
u1 = User(20)
del u1.age
```

### <b style='color:yellow;'> Interview Questions 


#### 1️⃣ What is encapsulation?

    Encapsulation is the process of wrapping data and methods together inside a class and restricting direct access to internal data.

#### 1️⃣ What is encapsulation?

    Encapsulation is the process of wrapping data and methods together inside a class and restricting direct access to internal data.

#### 2️⃣ Why is encapsulation important?

    Protects data from accidental modification
    
    Allows controlled access
    
    Improves maintainability
    
    Helps implement validation logic

#### 3️⃣ How is encapsulation implemented in Python?

    Using public, protected (_var), and private (__var) attributes and getter/setter methods.

#### 4️⃣ Public, Protected, Private in Python?
| Type      | Syntax     | Access                    |
| --------- | ---------- | ------------------------- |
| Public    | `self.x`   | Anywhere                  |
| Protected | `self._x`  | Internal use (convention) |
| Private   | `self.__x` | Name mangled              |


#### 5️⃣ Does Python support true private variables?

    No. Python uses name mangling, not strict access control.

#### 6️⃣ What is name mangling?

    Name mangling is when Python internally changes __var to _ClassName__var to prevent accidental access.

#### 7️⃣ Why does Python use name mangling?

    Avoid accidental access
    
    Prevent name conflict in inheritance
    
    Signal internal use

#### Is encapsulation for security?

    No. It is for design and protection from accidental misuse, not strict security.

#### Can private variables be accessed outside class?

    Yes, using _ClassName__var, but it’s discouraged.

#### How does encapsulation improve maintainability?

    Changes in internal logic don’t affect external code.

#### Encapsulation vs Abstraction?
| Encapsulation         | Abstraction          |
| --------------------- | -------------------- |
| Hides data            | Hides implementation |
| Focuses on protection | Focuses on structure |
