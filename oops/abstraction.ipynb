### <b>What is Abstraction?
    Abstraction means hiding unnecessary details and showing only essential features.

### <b>Real-World Analogy

#### ATM Machine

    When you use ATM:
    
        You insert card
        
        Enter PIN
        
        Withdraw money

#### Do you know:

    How banking server validates?
    
    How encryption works?
    
    How transaction logs work?

    ❌ No

You only see what you need.

    That is abstraction.

### <b>Technical Definition
    Abstraction is the process of hiding implementation details and exposing only necessary functionality.

### <b>Difference from Encapsulation
| Concept       | Focus                 |
| ------------- | --------------------- |
| Encapsulation | Hiding data           |
| Abstraction   | Hiding implementation |


### <b>How Abstraction is Implemented in Python

    to implement python provides the module called ABC

```python
from abc import ABC, abstractmethod
```

### <b>Step-by-Step Example

#### <b>Step 1: Create Abstract Class

```python
class Payment(ABC) : 

    @abstractmethod
    def pay(self, amount):
        pass
    
```

Here:

<i style="color:yellow;">Payment is an abstract class </i>
    
<i style="color:yellow;">pay() must be implemented by child classes </i>
    
<i style="color:yellow;">You cannot create object of Payment</i>

#### <b> Step 2: Child Classes Implement It

```python
class UPI(Payment): 
    def pay(self,amount): 
        print(f"paid {amount} via UPI")

class Card(Payment): 
    def pay(self,amount): 
        print(f"paid {amount} via Card")
```

### <b>Step 3: Use It

```python
upi = UPI()
upi.pay("2000")
```

```python
card = Card()
card.pay("3000")
```

### <b>Works ✅

```python
# But this will fail ❌
p = Payment()
```

```python
# Because abstract class cannot be instantiated.
```

### <b>What Abstraction Achieves

    ✔ Forces child classes to implement required methods
    ✔ Ensures structure
    ✔ Hides internal logic
    ✔ Makes system scalable

### <b>Key Rules of Abstraction in Python

    Use ABC class
    Use @abstractmethod decorator    
    Abstract class cannot be instantiated    
    Child must implement abstract methods
