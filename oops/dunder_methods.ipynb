### <b>What Are Dunder Methods?

    Dunder methods (Double UNDERscore methods) are special methods in Python that start and end with double underscores. 

# Example 
    __init__ 
    __str__ 
    __len__
    __add__

#### They are also called:

    Magic methods
    
    Special methods

### <b>Why Are They Special?

    Because Python automatically calls them.
    
    You don’t call them directly (usually).

### Example 1 <i> __init__ </i>

```python
class User: 
    def __init__(self, name): 
        self.name = name 
```

```python
When you do:
```

```python
u1 = User("yashwanth")
```

```python
python internally calls :
```

```python
User.__init__(u1,"Yashwanth")
```

    So __init__ is automatically called when object is created.

#### Example 2 — __str__

```python
class User: 
    def __init__(self,name): 
        self.name = name 

    def __str__(self):
        return f"My name is {self.name}"
```

```python
now: 
```

```python
u1 = User("Yashwanth")
print(u1)
```

```python
Python internally calls:
```

```python
u1.__str__()
```

    If __str__ is not defined, it prints memory address.

#### Example 3 — __add__

```python
class Number:
    def __init__(self, value):
        self.value = value

    def __add__(self, other):
        return self.value + other.value
```

```python
n1 = Number(100)
n2 = Number(200)
print(n1 + n2)
```

#### Example 4 — __len__

```python
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)
```

```python
t1 = Team("Yashwanth")
len(t1)
```

### __repr__
    __repr__ returns the official string representation of an object, mainly used for debugging and developers.

```python
class User:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def __repr__(self):
        return f"User('{self.name}', {self.age})"
```

```python
u1 = User("Yashwanth", 20)
print(u1)
```

### <b>Most Common Dunder Methods

| Method     | Purpose                   |
| ---------- | ------------------------- |
| `__init__` | Constructor               |
| `__str__`  | User-friendly string      |
| `__repr__` | Developer-friendly string |
| `__len__`  | Length                    |
| `__add__`  | + operator                |
| `__sub__`  | - operator                |
| `__mul__`  | * operator                |
| `__eq__`   | == comparison             |
| `__lt__`   | < comparison              |
| `__call__` | Make object callable      |
| `__iter__` | Make object iterable      |
| `__next__` | Iterator support          |


### <b style="color:yellow"> Interview Questions
Q1) What are dunder methods?

    Special methods with double underscores that define object behavior in Python.

Q2) Can we call dunder methods directly?

    Yes, but normally Python calls them automatically.

Q3) Why are they called magic methods?

    Because they allow objects to behave like built-in types.
