```python
class User: 
    
    def __init__(self,username,email): 
        self.username = username
        self.email = email

    def login(self):
        print(f"{self.username} is successfully logged")
```

```python
user1 = User("yashwanth","yashkothapalli@gmail.com")
user1.login()
```

### 1️⃣ Single Inheritance
🔸 Child: Admin

Admin is a User → perfect inheritance case.

```python
class Admin(User): 
    
    def delete_user(self):
        print(f"user deleted")

admin1 = Admin("yash","yashkothapalli")
admin1.login()
admin1.delete_user()
```

### 2️⃣ Multilevel Inheritance
🔸 SuperAdmin → inherits from Admin

```python
class SuperAdmin(Admin): 
 
    def access_logs(self):
        print(f"accessing system logs")

super_admin = SuperAdmin("yashwanth","yashkothapalli")
super_admin.login()
super_admin.delete_user()
super_admin.access_logs()
```

### 3️⃣ Hierarchical Inheritance
🔸 Another child of User

```python
class Customer(User):
    
    def purchase(self): 
        print(f"{self.username} product purchased")

customer = Customer("suraj","suraj@gmail,com")
customer.login()
customer.purchase()
```

### 4️⃣ Multiple Inheritance (IMPORTANT + TRICKY)
🔸 Extra Features

```python
class Logger: 
    def log_activity(self): 
        print(f"Activity logged")
        
```

- Child inherits from multiple parents

```python
class SupportUser(User,Logger): 
    def resolve_ticket(self): 
        print(f"ticket resolved")

support_user = SupportUser("mahesh","maheshvarma@gmail.com") 
support_user.login()
support_user.log_activity()
support_user.resolve_ticket()
```
