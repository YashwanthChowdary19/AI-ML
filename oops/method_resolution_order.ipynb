### <b>MRO — Method Resolution Order (Python)</b>
🔹 Definition 

Method Resolution Order (MRO) is the order in which Python searches for a method or attribute in a class hierarchy, especially when multiple inheritance is involved.

❓ Why MRO Exists

In multiple inheritance, the same method name may exist in more than one parent class.
Python needs a clear, consistent rule to decide:

“Which method should I call first?”

That rule is MRO.

### <b>Multiple Inheritance Example (REAL MRO CASE)<b>

```python
class A:
    def show(self):
        print("A")

class B:
    def show(self):
        print("B")

class C(A, B):
    pass

c = C()
c.show()
```

```python
MRO order : C -> A -> B -> object
```

### <b>How Python Calculates MRO<b>

Python uses the <b style=color:blue;>C3 Linearization Algorithm.<b>

📌 You don’t need to implement it — just remember:

Order is left to right

Parent order is preserved

No class appears twice

### <b>How to Check MRO

```python
syntax : 
classname dot mro()  # it returns as list
or 
classname dot __mro__  # it returns tuple
```

```python
# lets check the MRO for class "C"
C.mro()
```

```python
C.__mro__
```

### <b>💎 Diamond Problem

```python
        A
       / \
      B   C
       \ /
        D
# the nodes are classes 
```

```python
class A:
    def show(self):
        print("A")

class B(A):
    pass

class C(A):
    def show(self):
        print("C")

class D(B, C):
    pass

d = D()
d.show()
```

MRO order:
D → B → C → A → object
