📌 Definition:

Object-Oriented Programming is a programming paradigm where software is designed using objects that combine data (state,attributes) and behavior (methods).

🏠 Real-world analogy:

Class → Blueprint of a house

Object → Actual house built from blueprint

#### Class

A blueprint or template that defines the structure and behavior of objects.

#### Object

An instance of a class that represents a real-world entity with its own data and behavior.

```python
class User : 
    pass
```

class → keyword

User → class name (PascalCase)

pass → placeholder

```python
u1 = User()
```

u1 is an object (instance) of class User.

#### Attribute (Instance Variable)

A variable that stores data inside an object.

```python
u1.name = "yashwanth"
# name is an attribute
```

```python
u1.name
```

#### Method

A function that is defined inside a class and operates on object data.

```python
class User : 
    def login(self) : 
        print(f"login successful")
        
```

```python
u1 = User()
```

```python
u1.login()
# here login is the method
```

#### self

A reference to the current object that allows access to its attributes and methods.

```python
# lets check
class User : 
    def check(self) : 
        print(id(self))
u1 = User()
print(id(u1))
u1.check()
# both(self,u1) are pointing to the same memory
u2 = User()
print(id(u2))
u2.check()
```

#### Constructor (__init__)

A special method that is automatically executed when an object is created, used to initialize object data.

```python
class User : 
    def __init__(self) :
        print(f"constructor initialized")

u1 = User() # automatically called when object is created
```
