### <b>POLYMORPHISM
🔹 One-line meaning (remember this)

Polymorphism means “same method name, different behavior.”

That’s it.
If you remember only this, you already understand polymorphism.

🔹 Break the word (helps memory)

    Poly → many

    Morphism → forms

👉 Many forms of the same thing

### 🧑‍🏫 <b>Real-world analogy (BEST ONE)
Example: “Start” button

    In a car → starts engine

    In a mobile → starts app

    In a machine → starts process

    Same button name → different actions

That is polymorphism.

```python
# example code with same analogy 
class Car:
    def start(self): 
        print(f"start the engine")

class Mobile: 
    def start(self): 
        print(f"start the app")

class Machine: 
    def start(self): 
        print(f"start the process") 

car = Car()
car.start()
mobile = Mobile() 
mobile.start()
machine = Machine() 
machine.start()
```

```python
# exmaple-02 
class Dog: 
    def speak(self): 
        print("Dog barks")

class Cat:
    def speak(self): 
        print("cat meaws")

classes = [Dog(),Cat()]

for cls in classes: 
    cls.speak()
```

### <b>Key Point (IMPORTANT)

Python does not care about the class name.
It only cares:

“Does this object have a speak() method?”

This is called <h1><b style="color:blue;">duck typing.</h1>

### <b>METHOD OVERRIDING
            same name with same number of parameters in multiple classes
🔹 One-line meaning (remember this)

    Method overriding means: a child class replaces a parent class method with its own version.

    That’s all it is.

### 🏠 <b>Real-world analogy (VERY IMPORTANT)
Example: Rules in a company

    Company rule (Parent): 
    “Office opens at 9 AM”

    IT Department rule (Child): 
    “Office opens at 10 AM”

    Same rule name → different rule applies for IT.

👉 The child rule overrides the parent rule.

```python
class User: 
    def role(self): 
        print(f"General Role")

class Admin(User): 
    def role(self):
        print(f"Admin Role")

Admin().role()

# child class role function overrided by User class role function
```

#### <b>Python thinks:

    Does Admin have role()? ✅ YES
    Use it
    Ignore parent’s role()

####  <b>❌ Parent method is NOT deleted

    It is just hidden.
    You can still access it using super()

### <b>Method Overloading</b><b style="color:red;"> Not Supported in Python

    Method overloading means defining multiple methods with the same name but different parameters.

    Unlike languages such as Java, Python does not support true method overloading.
    If multiple methods with the same name are defined in the same class, Python keeps only the latest definition, and the previous ones are overwritten.

    but we can implement using default or variable length positional arguments

```python
# example implement like Java methodology
class Add: 
    def add(self,a,b): 
        return a+b 
    def add(self,a,b,c): 
        return a+b+c 
Add().add(2,3)   
```

```python
# to resolve above problem we use either default arguments for small number of arguments or variable length positional arguments  
# using default arguments
class Add: 
    def add(self,a,b,c=0): 
        return a+b+c 
Add().add(1,2,3)
```

    or

```python
# variable length positional arguments
class Add: 
    def add(self,*args): 
        count=0
        for i in args: 
            count += i 
        return count
Add().add(1,2,3,4,5,6,7)
```
