## <b>super()

### <b>What is super()?

    super() is used to call methods of the parent class inside a child class.

```python
# Example 
class Parent:
    def __init__(self):
        print("Parent constructor")

class Child(Parent):
    def __init__(self):
        super().__init__() # super method is used to call the parent constructor
        print("Child constructor")

c = Child()
```

### <b> Why super() is Important?

Without super():

    Parent constructor won’t run
    
    Important initialization may be skipped

In frameworks like Django, FastAPI — super() is heavily used.

### <b>Advanced: In Multiple Inheritance

    super() follows MRO (Method Resolution Order).

    It does NOT just call immediate parent.
    It follows C3 linearization order.

```python
# example 

class Parent: 
    def __init__(self): 
        print(f"parent constructor")

class Child(Parent):
    def __init__(self): 
        print(f"Child constructor")

class GrandChild(Child, Parent): 
    def __init__(self): 
        super().__init__() 
        print("Grand Child Constructor")

gc = GrandChild()
```

## <b>Class Variables vs Instance Variables

### 🔹 Instance Variable 

    belong to object 

```python
class User: 
    def __init__(self,name): 
        self.name = name  # name is an instance variable 
```

    Each object has separate copy of variable name

### 🔹 Class Variable

    belongs to class

```python
class User: 
    host = "Jupyter" # host is a class variable

```

    Shared among all objects.

### Important Difference

```python
class A:
    x = 10

a1 = A()
a2 = A()

a1.x = 20
```

```python
print(a1.x) 
print(a2.x)
print(A.x)
```

Why?

    Because a1.x = 20 creates instance variable, not modify class variable.
